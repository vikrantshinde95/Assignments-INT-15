### 12.	Generate the decorator for a Garden visit to check if current time is between 7am to 7pm.  If true the only call the function else generate the error.   a.
```python
from datetime import datetime

def check_time(func):
    def wrapper(*args, **kwargs):
        current_time = datetime.now().time()
        if current_time >= datetime.strptime("07:00:00", "%H:%M:%S").time() and current_time <= datetime.strptime("19:00:00", "%H:%M:%S").time():
            return func(*args, **kwargs)
        else:
            raise Exception("Function can only be called between 7 am and 7 pm")
    return wrapper

@check_time
def payment_check(user_transactions):
    print("Checking payment transactions...")

@check_time
def entry(user_id):
    print(f"User {user_id} entered the garden.")

@check_time
def exit(user_id):
    print(f"User {user_id} exited the garden.")
```

#### 18.	Segregate file in small , medium and large size in different folder
#### o	filesize less than and equal to 10KB is small file
#### o	filesize more than 10 KB is the medium file
#### o	filesize more than 40 kb is the large file
#### o	create the log file 


```python
import os
import shutil
import logging
```

```python

def seperate_files(source_folder):
    
    small_folder = os.path.join(source_folder, "small_files")
    medium_folder = os.path.join(source_folder, "medium_files")
    large_folder = os.path.join(source_folder, "large_files")

    os.makedirs(small_folder, exist_ok=True)
    os.makedirs(medium_folder, exist_ok=True)
    os.makedirs(large_folder, exist_ok=True)

    logging.basicConfig(filename='log.txt', level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

    for filename in os.listdir(source_folder):
        file_path = os.path.join(source_folder, filename)

        if os.path.isdir(file_path):
            continue
        
        file_size = os.path.getsize(file_path)  
        file_size_kb = file_size / 1024 

        if file_size_kb <= 10:
            destination_folder = small_folder
            logging.info(f"small file_name: {filename}, Size: {file_size_kb:.2f} KB")
        elif 10 < file_size_kb <= 40:
            destination_folder = medium_folder
            logging.info(f"medium file_name: {filename}, Size: {file_size_kb:.2f} KB")
        else:
            destination_folder = large_folder
            logging.info(f"large_file: {filename}, Size: {file_size_kb:.2f} KB")

        shutil.move(file_path, os.path.join(destination_folder, filename))

source_folder = "diiferent_size_folders"  

seperate_files(source_folder)

print("Files have been seperated and the log file has been created.")
```

```python

def most_frequent_word(file_path):
    if not os.path.exists(file_path):
        print("The file does not exist.")
    
    try:
        with open(file_path, 'r') as file:
            content = file.read().lower()
            words = ''.join(char if char.isalnum() else ' ' for char in content).split()
            
            word_frequency = {}
            for word in words:
                if word in word_frequency:
                    word_frequency[word] += 1
                else:
                    word_frequency[word] = 1
            
            most_common_word = max(word_frequency, key=word_frequency.get)
            print(f"The most frequent word is '{most_common_word}' with {word_frequency[most_common_word]} occurrences.")
    
    except Exception as e:
        print(f"An error occurred: {e}")


```

```python
class FileReader:
    def __init__(self, file_path):
        self.file_path = file_path
        self.numbers_dict = {}

    def read_file(self):
        with open(self.file_path, 'r') as file:
            content = file.read()
            self.extract_numbers(content)

    def extract_numbers(self, content):
        words = content.split()
        for i, word in enumerate(words):
            if word.isdigit():
                self.numbers_dict[i] = int(word)

    def get_numbers_dict(self):
        return self.numbers_dict


class User:
    def __init__(self, name, file_path):
        self.name = name
        self.file_reader = FileReader(file_path)
        self.numbers_dict = {}

    def process_file(self):
        self.file_reader.read_file()
        self.numbers_dict = self.file_reader.get_numbers_dict()

    def get_user_data(self):
        return {self.name: self.numbers_dict}

```

```python
class User:
    def __init__(self, username):
        self.username = username

    def register(self):
        print(f"{self.username} registered.")

    def sign_in(self):
        print(f"{self.username} signed in.")

    def sign_out(self):
        print(f"{self.username} signed out.")


class Student(User):
    def __init__(self, username):
        super().__init__(username)
        self.marks = None
        self.result_date = None

    def view_marks(self):
        if self.marks and self.result_date and date.today() >= self.result_date:
            print(f"{self.username}'s marks: {self.marks}")
        else:
            print("Marks are not available yet.")


class Professor(User):
    def enter_marks(self, student, marks, result_date):
        student.marks = marks
        student.result_date = result_date
        print(f"Marks entered for {student.username}.")


class Admin(User):
    def view_marks(self, student):
        if student.marks:
            print(f"{student.username}'s marks: {student.marks}")
        else:
            print("Marks are not available yet.")

```

```python
admin = Admin("admin")
admin.register()
admin.sign_in()
```

```python
professor = Professor("professor")
professor.register()
professor.sign_in()
```

```python
student = Student("student")
student.register()
student.sign_in()
```

```python
professor.enter_marks(student, 85, date(2025, 2, 20))
```

```python
admin.view_marks(student)
student.view_marks()
```

```python
student.result_date = date.today()
student.view_marks()
```

#### 34.Create suitable classes, methods and attributes for resume exchange. Employee to upload the file and the same file is visible to employer 

```python
class ResumeManager:
    def __init__(self):
        self.resumes = {}

    def upload_resume(self, employee_name, file_name):
        self.resumes[employee_name] = file_name
        print(f"{employee_name} uploaded the resume: {file_name}")

    def view_resumes(self):
        print("Uploaded Resumes:")
        for employee, file_name in self.resumes.items():
            print(f"{employee}: {file_name}")
```

```python
manager = ResumeManager()
```

```python
manager.upload_resume("Vikrant", "alice_resume.pdf")
```

```python
manager.view_resumes()
```

#### Create the proper class with methods like user information, calculate the salary, apply for the leave, view leaves, approve leaves. Records for leave application should be stored in dummy table (temporary view). Once the leave is approved then it should move to permanent table.

```python
import pyodbc

from datetime import date
```

```python
server = 'VICKY'
database = 'biz'
connection_string = f'DRIVER={{SQL SERVER}};SERVER={server};DATABASE={database};'
```

```python
class Employee:
    def __init__(self, name, mobile, address):
        self.name = name
        self.mobile = mobile
        self.address = address
        self.id = self.save_to_db()

    def save_to_db(self):
        conn = pyodbc.connect(connection_string)
        cursor = conn.cursor()
        cursor.execute("INSERT INTO employees (name, mobile, address) OUTPUT INSERTED.id VALUES (?, ?, ?)",
                       (self.name, self.mobile, self.address))
        employee_id = cursor.fetchone()[0]
        conn.commit()
        conn.close()
        return employee_id

    def calculate_salary(self, base_salary, days_worked):
        salary = base_salary * days_worked
        return salary

    def apply_leave(self, leave_type, start_date, end_date):
        conn = pyodbc.connect(connection_string)
        cursor = conn.cursor()
        cursor.execute("INSERT INTO dummy_leaves (employee_id, leave_type, start_date, end_date, status) VALUES (?, ?, ?, ?, ?)",
                       (self.id, leave_type, start_date, end_date, 'Pending'))
        conn.commit()
        conn.close()

    def view_leaves(self):
        conn = pyodbc.connect(connection_string)
        cursor = conn.cursor()
        cursor.execute("SELECT * FROM leaves WHERE employee_id = ?", (self.id,))
        leaves = cursor.fetchall()
        conn.close()
        return leaves

class Admin:
    def approve_leave(self, leave_id):
        conn = pyodbc.connect(connection_string)
        cursor = conn.cursor()
        
        cursor.execute("SELECT * FROM dummy_leaves WHERE id = ?", (leave_id,))
        leave = cursor.fetchone()
        if leave:
            cursor.execute("INSERT INTO leaves (employee_id, leave_type, start_date, end_date, status) VALUES (?, ?, ?, ?, ?)",
                           (leave[1], leave[2], leave[3], leave[4], 'Approved'))
            cursor.execute("DELETE FROM dummy_leaves WHERE id = ?", (leave_id,))
        
        conn.commit()
        conn.close()

```

```python

salary = employee.calculate_salary(500, 20)
print(f"Salary for {employee.name}: {salary}")
```
